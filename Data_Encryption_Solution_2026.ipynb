### 1. Import libraries

```python
import pandas as pd
import hmac
import hashlib
import secrets
import os
import getpass
from pathlib import Path
```

### 2. Generate a secret key

```python
def generate_secret_key():

    return secrets.token_bytes(32)
```

### 3. Create the pseudonymisation function

```python
def generate_encrypted_id(identifier, secret_key):

    if pd.isna(identifier):
        return None

    identifier = str(identifier).strip()

    if not identifier:
        return None

    token = hmac.new(
        secret_key,
        identifier.encode("utf-8"),
        hashlib.sha256
    ).hexdigest()

    return token
```

### 4. Add the encrypted ID column

```python
def add_encrypted_id(df, id_column, secret_key):

    if id_column not in df.columns:
        raise ValueError(f"Column '{id_column}' not found.")

    df = df.copy()

    df["encrypted_id"] = df[id_column].apply(
        lambda x: generate_encrypted_id(x, secret_key)
    )

    return df
```

### 5. Test using synthetic data

```python
secret_key = generate_secret_key()

data1 = pd.DataFrame({
    "id": ["1001", "1002", "1003"],
    "name": ["John", "Mary", "Peter"]
})

data2 = pd.DataFrame({
    "id": ["1001", "1003", "1004"],
    "age": [45, 51, 38]
})

df1 = add_encrypted_id(data1, "id", secret_key)
df2 = add_encrypted_id(data2, "id", secret_key)

display(df1)
display(df2)
```

### 6. Verify consistency

```python
common_ids = set(df1["encrypted_id"]) & set(df2["encrypted_id"])

print("Matching pseudonymous IDs:", len(common_ids))
```
