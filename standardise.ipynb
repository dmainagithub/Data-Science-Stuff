### Standardization Solution

Notes: This function helps you ton standardize columns (either to capital case or small case)
Just point to the file you want processed and indicate the output directory

```python

import pandas as pd
from pathlib import Path

# Input and output files
input_file = Path("/dataA/madiva/data/dmac/nairobidb/hf_individuals_nairobi.csv") # "input.csv"
output_file = Path("/home/dmaina/pprl/output/hf_individuals_nairobi.csv")    # "input_uppercase.csv"

# Read CSV
df = pd.read_csv(input_file)

# Convert all text columns to uppercase
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].str.strip().str.upper()

# Save transformed data
df.to_csv(output_file, index=False)

print("Transformation completed!")
print(f"Original records: {len(df):,}")
print(f"Output file: {output_file}")
print(f"Columns: {list(df.columns)}")
```
